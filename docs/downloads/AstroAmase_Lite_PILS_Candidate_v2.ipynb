## Transition formats

The CSV schema is `astroamase-local-csv-v1`. Its exact header is

```csv
frequency_MHz,uncertainty_MHz,A_s-1,E_upper_K,g_upper,qn_upper,qn_lower
```

Supply a positive rest frequency, nonnegative frequency uncertainty, positive Einstein A in inverse seconds, upper-state energy above the declared zero in kelvin, positive integer upper-state degeneracy and both level labels. Quote CSV labels containing commas. Equal-frequency components can coexist when their level labels distinguish them. Rows are sorted by frequency for the fitter. The lower-state energy follows from the upper energy and photon frequency.

The `astroamase-local-spcat-v1` schema accepts a restricted SPCAT format in ASCII. Declare the source file's absolute species tag in `spcat_tag`, separately from the local run tag. Set `intensity_convention` to `log10_nm2_MHz_at_300_K` only for decadic integrated intensity in nm² MHz at 300 K. Each row uses the standard widths 13, 8, 8, 2, 10, 3, 7 and 4 for frequency in MHz, uncertainty in MHz, log intensity, rotational degrees of freedom, lower-state energy in inverse centimeters, decimal upper degeneracy, signed source tag and QNFMT. The two 12-character quantum-number fields follow, with two characters per slot. The loader preserves the occupied QN tokens. A signed source tag is accepted for measured lines. Overflow markers, encoded degeneracies, unsupported QN layouts, negative frequency uncertainties and alternate intensity units require conversion from the authoritative source to the explicit CSV schema.

The SPCAT conversion uses the upstream `Catalog._set_sijmu_aij` coefficients. With frequency in MHz, lower energy in K, and 300 K intensity I, it gives

```
A = (2.40251e4 × 1.16395e−20) × I × Q(300) × frequency²
    / [g_upper × exp(−E_lower/300) × (1 − exp(−hν/k/300))]
```

Here ν in the photon energy is in Hz. The conversion evaluates the small exponential difference using `expm1`. In CSV mode the inverse expression supplies log intensity for plotting. The simulator uses the supplied Einstein A directly.

A single input is limited to 100,000 transitions and 16 MiB of transition text. A run permits 32 additions totaling up to 64 MiB. A metadata file is limited to 64 KiB. Use the template's fixed file names in a saved catalog folder. The staging function accepts other original base names with the matching transition extension.

## 4. Review the fit settings

The telescope geometry and source diameter affect the line intensities. Enter a dish diameter for single-dish data or both beam axes for interferometric data. A fixed VLSR requires an exact excitation temperature. To estimate VLSR, clear the fixed velocity field, choose whether to fix or estimate temperature, and set the velocity search bounds. Leave linewidth or RMS blank to ask Astro AMASE to estimate it. Automatic estimates with this draft catalog scope have limited validation. Choose `0` fitting iterations to run until convergence.

Leave **valid_elements** blank for the upstream C, O, H, N and S setting. An atom outside the list halves the candidate’s global score. It also removes the candidate from the clean unassigned alternatives in diagnostics. Enter case-sensitive symbols separated by commas, such as `C,O,H,N,S,Si`, or `all` for all 118 elements. A LOCAL species with no SMILES uses its declared element list.

When estimating VLSR, leave **vlsr_mols** blank for the curated tracers or enter exact active catalog labels as a JSON list, such as `["H2CO", "CH3CN, v = 0"]`. **known_molecules** takes a JSON list of SMILES for molecules independently known in the source.
